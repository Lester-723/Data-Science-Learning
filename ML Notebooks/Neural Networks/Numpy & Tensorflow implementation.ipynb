```python
def greeting_function(func):

    def wrapper(a_in,w,b):
        print("The Dense function has started")
        result = func(a_in,w,b)
        print("The Dense function has ended")
        return result
    return wrapper
```

```python
def activation_function(z):
    return 1/ ( 1 + np.exp(-z))
    
```

```python
np.zeros((3,2))[:,0]
```

```python
@greeting_function
def dense_function(a_in,w,b):
    """
    neuron_units - from the shape of the weight matrix the neurons are determined
    a_out - based on the multiplicative property of the matrix the shape of the output matrix is determined
    i_out - individual output of a singular neuron
    z = is the output of the neuron
    """
    neuron_units = a_in.shape[0] 
    neuron_output = w.shape[1]
    a_out = np.zeros((neuron_units,neuron_output))
    for j in range(neuron_output):
        i = w[:,j]
        z = np.dot(a_in,i) + b[j]
        a_out[:,j] = activation_function(z)
    return a_out

dense_function(a_in = x_matrix,w = w_matrix,b = b_matrix)

```

```python
def sequential(a_in,w,b,w1,b1):
    a1 = dense_function(a_in , w , b)
    a2 = dense_function(a1,w1,b1)
    return a2
```

```python
prediction = sequential(a_in = x_matrix , w = w_matrix , b = b_matrix , w1 = w1_matrix , b1 = b1_matrix)
```

```python
def predict(predictions, threshold):
    y_pred = np.zeros_like(predictions)
    for i in range(len(predictions)):
        if predictions[i] >= threshold:
            y_pred[i] = 1
        else:
            y_pred[i] = 0
    return y_pred
        
    
```

```python
predict(predictions= prediction,threshold=0.3)
```
